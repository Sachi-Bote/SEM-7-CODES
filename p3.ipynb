```python
import numpy as np
import matplotlib.pyplot as plt
import time
```

```python
# 1. Define the function and its derivative
def function(x):
    return (x + 3) ** 2

def gradient(x):
    return 2 * (x + 3)
```

```python
# 2. Initialize parameters
x = 2
learning_rate = 0.1
iterations = 30

# Store values for plotting
x_values = [x]
y_values = [function(x)]
```

```python
# 3. Apply Gradient Descent
start_time = time.time()

print("Iteration\t x\t\t y")

for i in range(iterations):
    x = x - learning_rate * gradient(x)

    x_values.append(x)
    y_values.append(function(x))

    print(f"{i+1}\t\t {x:.6f}\t {function(x):.6f}")

execution_time = time.time() - start_time
```

```python
# 4. Display minimum and execution time
print("\nLocal Minimum:")
print("x =", round(x, 6))
print("y =", round(function(x), 6))
print("Execution Time:", round(execution_time, 6), "seconds")
```

```python
# 5. Plot the function and descent path
x_plot = np.linspace(-5, 3, 100)
y_plot = function(x_plot)

plt.plot(x_plot, y_plot, label="y = (x+3)^2")
plt.scatter(x_values, y_values, color='red', label="Gradient Descent")
plt.scatter(x_values[0], y_values[0], color='green', label="Starting Point")
plt.scatter(x_values[-1], y_values[-1], color='blue', label="Minimum")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent to Find Local Minimum")
plt.legend()
plt.grid()
plt.show()
```
